```python
import pandas as pd
import matplotlib.pyplot as plt

print("Second Avenue Subway analysis environment is ready!")
```

```python
proposals = pd.DataFrame({
    "year": [1929, 1944, 1968, 2007, 2017],
    "proposal": [
        "Initial IND Second Avenue Subway proposal",
        "Board of Transportation expansion proposal",
        "Program for Action proposal",
        "Phase 1 construction restarted",
        "Phase 1 opened"
    ]
})

proposals
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,4))

plt.scatter(proposals["year"], [1]*len(proposals), s=120)

for i, row in proposals.iterrows():
    plt.text(
        row["year"],
        1.02,
        row["proposal"],
        rotation=45,
        ha="right"
    )

plt.yticks([])
plt.xlabel("Year")
plt.title("Second Avenue Subway Timeline") 

plt.savefig("../visuals/second_avenue_subway_timeline.png", bbox_inches="tight", dpi=300)

plt.show()
```
